```python
exclusions_bleed = [
    "./exclusions_bleed_gh.csv",
    "./exclusions_bleed_yk.csv",
]

exclusions_non_bleed = [
    # "./exclusions_no_bleed_gh.csv",
    "./exclusions_no_bleed_yk.csv",
]

columns = ['ct_axial', 'mri_b0_axial', 'mri_b1000_axial', 'adc']
```

```python
# load csvs
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

def load_exclusions(file_list):
    df_list = [pd.read_csv(f) for f in file_list]
    df = pd.concat(df_list, ignore_index=True)
    return df

df_bleed = load_exclusions(exclusions_bleed)
df_non_bleed = load_exclusions(exclusions_non_bleed)

# change formats
for col in columns:
    df_bleed[col] = './skull_strip/bleed_ct_mri/' + df_bleed['case'] + '/' + col + '/' + df_bleed['index'].map(lambda x: f"{x:04d}") + '.jpg'
    df_non_bleed[col] = './skull_strip/no_bleed_ct_mri/' + df_non_bleed['case'] + '/' + col + '/' + df_non_bleed['index'].map(lambda x: f"{x:04d}") + '.jpg'

# remove duplicates
df_bleed = df_bleed.drop_duplicates()
df_non_bleed = df_non_bleed.drop_duplicates()

print(f"Bleed exclusions: {len(df_bleed)}")
print(f"Non-bleed exclusions: {len(df_non_bleed)}")
```

```python
# 경로 구조 확인
print("=== Directory Structure Check ===")
count = 0
for (path, dirs, files) in os.walk('./skull_strip/bleed_ct_mri'):
    png_files = [f for f in files if f.endswith('.png') and 'checkpoint' not in f]
    if png_files:
        print(f"Path: {path}")
        print(f"  PNG files: {len(png_files)}")
        print(f"  Example: {png_files[0]}")
        print(f"  Path parts: {path.split('/')}")
        count += 1
        if count >= 3:  # 처음 3개만 확인
            break
```

```python
import shutil
transferred_files = 0
excepted_files = 0

for (path, dir, files) in os.walk('./skull_strip/bleed_ct_mri'):
    for file in files:
        if file.endswith('.png') and 'checkpoint' not in file:
            path_parts = path.split('/')
            case = path_parts[-2]
            ct_type = path_parts[-1]
            
            # adc continue 제거
            index = int(file.split('.')[0])
            
            if not ((df_bleed['case'] == case) & (df_bleed['index'] == index)).any():
                transferred_files += 1
                src = os.path.join(path, file)
                dst_dir = os.path.join('./processed_with_skullstrip/bleed', case, ct_type)
                os.makedirs(dst_dir, exist_ok=True)
                dst = os.path.join(dst_dir, file)
                shutil.copy(src, dst)
            else:
                excepted_files += 1

print(f"Transferred files: {transferred_files}")
print(f"Excepted files: {excepted_files}")
```

```python
import shutil
transferred_files = 0
excepted_files = 0

for (path, dir, files) in os.walk('./skull_strip/no_bleed_ct_mri'):
    for file in files:
        if file.endswith('.png') and 'checkpoint' not in file:
            path_parts = path.split('/')
            case = path_parts[-2]
            ct_type = path_parts[-1]
            
            # adc continue 제거
            index = int(file.split('.')[0])
            
            if not ((df_non_bleed['case'] == case) & (df_non_bleed['index'] == index)).any():
                transferred_files += 1
                src = os.path.join(path, file)
                dst_dir = os.path.join('./processed_with_skullstrip/non_bleed', case, ct_type)
                os.makedirs(dst_dir, exist_ok=True)
                dst = os.path.join(dst_dir, file)
                shutil.copy(src, dst)
            else:
                excepted_files += 1

print(f"Transferred files: {transferred_files}")
print(f"Excepted files: {excepted_files}")
```
